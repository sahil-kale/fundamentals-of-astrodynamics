For a certain satellite at $\nu = 90°$, the observed velocity and radius are:

$$v = 45{,}000 \; \text{ft/s}, \quad r = 4{,}000 \; \text{n mi}$$

Find the **eccentricity** of the orbit.

*Hint: at $\nu = 90°$, $r = p/(1+e\cos90°) = p$, so $p = r$.  Then use the vis-viva / h approach.*

*(Answer: e = 1.581)*

```python
# ── Problem 1.2 ───────────────────────────────────────────────────────────────
v_fps  = 45_000          # ft/s
r_nmi  = 4_000           # n mi
r_ft   = r_nmi * NMI_TO_FT

# Your work here:
# we know that r = p/(1 + e cos (nu)), at 90 degrees r = p.
# then we can solve for h
p = r_ft
h = np.sqrt(p * MU_FT)
specific_mechanical_energy = v_fps * v_fps/2 - MU_FT/r_ft
a = -MU_FT / (2 * specific_mechanical_energy)
e = np.sqrt(1 - p/a)

print(f"e = {e}")
```

## Problem 1.8
Identify each trajectory as **circular, elliptical, parabolic, or hyperbolic**:

| | Given | Identify |
|---|---|---|
| a | r = 3 DU, v = 1.5 DU/TU | ? |
| b | r_perigee = 1.5 DU, p = 3 DU | ? |
| c | ε = −1/3 DU²/TU², p = 1.5 DU | ? |
| d | **r** = Ĵ + 0.2K̂, **v** = 0.9Î + 0.123K̂ | ? |
| e | **r** = 1.01K̂, **v** = Î + 1.4K̂ | ? |

```python
# ── Problem 1.8 ───────────────────────────────────────────────────────────────
# Strategy: compute eccentricity or specific energy; classify from sign of ε.
#   ε < 0  → ellipse (or circle if e=0)
#   ε = 0  → parabola
#   ε > 0  → hyperbola
# string: 'circular','elliptical','parabolic','hyperbolic'

# (a)
r_a = 3.0; v_a = 1.5
specific_mechanical_energy = v_a**2/2 - MU_CANON/r_a
print(f"specific_mechanical_energy: {specific_mechanical_energy}")
type_a = "hyperbolic"

# (b)
r_peri_b = 1.5; p_b = 3.0
e_b    = p_b/r_peri_b - 1
print(f"e_b: {e_b}")
type_b = "hyperbolic"

# (c)
eps_c = -1/3; p_c = 1.5
a = -MU_CANON/(2* eps_c)
e_c    = np.sqrt(1-p_c/a)
print(f"e_c: {e_c}")
type_c = "circular"

# (d)
r_d = np.array([0.0, 1.0, 0.2])
v_d = np.array([0.9, 0.0, 0.123])
h = np.cross(r_d, v_d)
e_d   = np.cross(v_d, h)/MU_CANON - r_d/np.linalg.norm(r_d)
print(f"e_d: {np.linalg.norm(e_d)}")
type_d = "elliptical"

# (e)
r_e = np.array([0.0, 0.0, 1.01])
v_e = np.array([1.0, 0.0, 1.4])
h = np.cross(r_e, v_e)
e_e   = np.cross(v_e, h)/MU_CANON - r_e/np.linalg.norm(r_e)
print(f"e_e: {np.linalg.norm(e_e)}")
type_e = "hyperbolic"

for lbl, t in zip('abcde', [type_a, type_b, type_c, type_d, type_e]):
    print(f"({lbl}) {t}")
```

```python
# ── Answer check 1.8 ─────────────────────────────────────────────────────────
def classify_energy(eps):
    if abs(eps) < 1e-6: return 'parabolic'
    return 'elliptical' if eps < 0 else 'hyperbolic'

def energy_from_rv(r_vec, v_vec, mu=MU_CANON):
    r = np.linalg.norm(r_vec)
    v = np.linalg.norm(v_vec)
    return v**2/2 - mu/r

# Reference answers
eps_a_ref = v_a**2/2 - MU_CANON/r_a
# circ speed at r=3: v_c = sqrt(mu/r) = sqrt(1/3) ~ 0.577
v_circ_a = np.sqrt(MU_CANON / r_a)  # ~0.577; v_a=1.5 > v_esc=sqrt(2)*0.577~0.816 → hyperbolic

eps_b_ref = None  # need a to get energy; alternatively e from r_peri and p
# p = a(1-e^2) and r_peri = a(1-e) → e = p/r_peri - 1
e_b_ref = p_b / r_peri_b - 1

eps_d_ref = energy_from_rv(r_d, v_d)
eps_e_ref = energy_from_rv(r_e, v_e)

answers = {
    'a': ('hyperbolic',  type_a),
    'b': ('elliptical' if e_b_ref < 1 else 'hyperbolic', type_b),  # e=1 → parabolic
    'c': ('circular',  type_c), 
    'd': (classify_energy(eps_d_ref), type_d),
    'e': (classify_energy(eps_e_ref), type_e),
}
for lbl, (expected, got) in answers.items():
    status = "✅" if got == expected else "❌"
    print(f"{status} ({lbl})  expected={expected}, got={got}")
print(f"\nReference: e_b = {e_b_ref:.3f}, ε_d = {eps_d_ref:.4f}, ε_e = {eps_e_ref:.4f}")
```

## Problem 1.9
A space vehicle enters the sensible atmosphere at **300,000 ft** altitude with:
- Speed $v_1 = 25{,}000$ ft/s  
- Flight-path angle $\phi_1 = -60°$

Find its speed and flight-path angle at **100 n mi altitude** during descent.

*(Answer: V = 24,618 ft/s, φ = −59° 58')*

*Hint: conserve energy and angular momentum between the two points.*

```python
# ── Problem 1.9 ───────────────────────────────────────────────────────────────
h1_ft   = 300_000          # altitude at entry, ft
v1      = 25_000.0         # ft/s
phi1    = np.radians(-60)  # flight-path angle, rad

h2_alt  = 100 * NMI_TO_FT  # altitude at 100 n mi, ft

r1 = R_EARTH_FT + h1_ft
r2 = R_EARTH_FT + h2_alt

# Your work here (conserve ε and h_mag):
specific_angular_momentum = r1 * v1 * np.cos(phi1)
specific_mechanical_energy = v1**2/2 - MU_FT/r1

v2   = np.sqrt(2*(specific_mechanical_energy + MU_FT/r2))
phi2 = np.arccos(specific_angular_momentum/(r2*v2)) * -1 # vehicle is descending, phi must be negative

print(f"v2   = {v2:.1f} ft/s")
print(f"phi2 = {np.degrees(phi2):.4f}°" if phi2 is not None else "phi2 = None")
```

```python
# ── Answer check 1.9 ─────────────────────────────────────────────────────────
check("v2 (ft/s)",              v2,   24_618.0)
phi2_deg_ref = -(59 + 58/60)
if phi2 is not None:
    check("phi2 (deg)", np.degrees(phi2), phi2_deg_ref, rtol=5e-3)
```

Show that two-body motion is confined to a **plane fixed in space**.

*Hint: Show that **h** = **r** × **v** is a constant vector, so **r** and **v** always lie in the plane perpendicular to **h**.*

**Your proof for 1.10:**

> *(edit here)*

```python
# ── Hint / check 1.10 ────────────────────────────────────────────────────────
print("""
Proof outline:
1. Equation of motion: r̈ = −(μ/r³)r  (gravity is central — always along r̂)
2. Take d/dt(r × ṙ):
       d/dt(r × ṙ) = ṙ × ṙ + r × r̈ = 0 + r × (−μ/r³)r = 0
   because r × r = 0 for any vector.
3. Therefore h = r × v is constant (magnitude AND direction).
4. Since r · h = r · (r × v) = 0 always, r lies in the plane ⊥ to h.
   Similarly v · h = 0.  Both r and v are always in the same fixed plane. ✓
""")
```

**(c)** What is the physical significance of part (b)?

**Your derivation for 1.15:**

> *(edit here)*

```python
print("""
(a) Summing all N equations of motion, the internal gravitational forces
    cancel in pairs (Newton's 3rd law: F_jk = −F_kj), leaving
    Σ m_k r̈_k = 0.

(b) r_c = Σ(m_k r_k) / M  →  M r̈_c = 0
    Integrating twice:  r_c = a·t + b  (a, b constant vectors)

(c) The system's center of mass moves at constant velocity in a straight
    line (or remains stationary) — i.e., there is no net external force.
    This is conservation of linear momentum for an isolated system.
""")
```

## Problem 1.21 ⭐ *(Derivation — repulsive force)*
Two spherically symmetric bodies interact via a **repulsive** force:
$$\mathbf{F} = +\frac{K m_1 m_2}{r^3}\hat{r}$$
(proportional to the product of masses, inversely proportional to cube of distance, repulsive — acting along the line connecting centers).

Assuming Newton's 2nd law holds, derive the **differential equation of motion**.

**Your derivation for 1.21:**

> *(edit here)*

```python
print("""
Let r = r₁ − r₂ be the relative position vector.

For body 1:   m₁ r̈₁ = +K m₁m₂ / r³  r̂   (repulsive, away from body 2)
For body 2:   m₂ r̈₂ = −K m₁m₂ / r³  r̂   (reaction, toward body 1)

Subtract:
   r̈₁ − r̈₂ = K m₂/r³ r̂ + K m₁/r³ r̂  = K(m₁+m₂)/r³ · r̂

Since r = |r₁ − r₂| and r̂ = r/r:

   r̈ = +K(m₁+m₂) / r³ · r

Compare to gravity:  r̈ = −G(m₁+m₂)/r³ · r
The repulsive case flips the sign.  The equation of motion is:

   r̈ − [K(m₁+m₂)/r³] r = 0

(This produces hyperbolic trajectories with the bodies deflecting away.)
""")
```

## Problem 1.22 ⭐ *(Mars mission parking orbit)*
A space vehicle is first launched into a **100 n mi circular parking orbit**.
Then an impulsive burn increases velocity by **37,600 ft/s** placing it on an **interplanetary trajectory** toward Mars.

**(a)** Speed in the parking orbit  
**(b)** $e$, $h$, and $\mathcal{E}$ for the escape orbit. What type of orbit?  
**(c)** Compare velocity at $10^6$ n mi from Earth with hyperbolic excess velocity $v_\infty$

```python
# ── Problem 1.22 ──────────────────────────────────────────────────────────────
alt_park  = 100 * NMI_TO_FT         # ft
r_park    = R_EARTH_FT + alt_park   # ft
delta_v   = 37_600.0                # ft/s

# (a) Circular speed in parking orbit
specific_mechanical_energy_init = -MU_FT/(2*r_park)
v_circ = np.sqrt(2*(specific_mechanical_energy_init + MU_FT/r_park))

# (b) After burn
v_esc_orbit = v_circ+delta_v   # ft/s 
energy_b    = v_esc_orbit**2/2 - MU_FT/r_park
h_b         = r_park * v_esc_orbit * np.cos(0)   # ft²/s  (circular orbit → v perpendicular to r)
e_b         = np.sqrt(1 + (2*energy_b * h_b**2)/(MU_FT**2))
print(f"e_b: {e_b}")
orbit_type  = "hyperbolic"

# (c) Compare at 1e6 n mi
r_far     = 1e6 * NMI_TO_FT + R_EARTH_FT   # ft
v_at_far  = np.sqrt(2*(energy_b + MU_FT/r_far))  # ft/s  (from energy conservation)
v_inf     = np.sqrt(2*energy_b)   # ft/s  hyperbolic excess velocity = sqrt(2ε)

print(f"(a) v_circ     = {v_circ:.1f} ft/s")
print(f"(b) energy     = {energy_b:.4e} ft²/s²")
print(f"(b) h          = {h_b:.4e} ft²/s")
print(f"(b) e          = {e_b:.4f}")
print(f"(b) orbit type = {orbit_type}")
print(f"(c) v at 1e6 nmi = {v_at_far:.1f} ft/s")
print(f"(c) v_inf        = {v_inf:.1f} ft/s")
```

```python
# ── Answer check 1.22 ────────────────────────────────────────────────────────
v_c_ref   = np.sqrt(MU_FT / r_park)
v_b_ref   = v_c_ref + delta_v
eps_ref   = v_b_ref**2/2 - MU_FT/r_park
h_ref     = r_park * v_b_ref   # circular parking → v perp to r, h = r*v_tangential
# Strictly: after burn still at same point, but now v is tangential (was circular)
e_ref     = np.sqrt(1 + 2*eps_ref*h_ref**2/MU_FT**2)
r_far_ref = 1e6 * NMI_TO_FT + R_EARTH_FT
v_far_ref = np.sqrt(2*(eps_ref + MU_FT/r_far_ref))
v_inf_ref = np.sqrt(2*eps_ref) if eps_ref > 0 else None

check("(a) v_circ (ft/s)",     v_circ,    v_c_ref)
check("(b) energy (ft²/s²)",   energy_b,  eps_ref)
check("(b) e",                  e_b,       e_ref)
check("(c) v at 1e6 nmi (ft/s)", v_at_far, v_far_ref)
if v_inf_ref:
    check("(c) v_inf (ft/s)",   v_inf,     v_inf_ref)
    print(f"\nv_far/v_inf = {v_far_ref/v_inf_ref:.6f}  (should be ≈ 1.0)")
```
